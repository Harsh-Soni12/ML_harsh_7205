```python
import pandas as pd
import numpy as np
from google.colab import files

#uploaded = files.upload()
df = pd.read_csv('00 Social_Network_Ads.csv')

df.head()

```

```python
x = df[['Age','EstimatedSalary']]
y = df['Purchased']


print("--- Features (x) ---")
display(x.head())

print("\n--- Target (y) ---")
display(y.head())
```

```python
from sklearn.model_selection import train_test_split

```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=1)
```

```python
from sklearn.preprocessing import StandardScaler

```

```python
ss = StandardScaler().set_output(transform='pandas')
```

```python
x_train_scaled = ss.fit_transform(x_train)
x_test_scaled = ss.transform(x_test)
```

```python
x_train.describe()
```

```python
x_train_scaled.describe()
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()

lr_scaled = LogisticRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
lr_scaled.fit(x_train_scaled, y_train)
```

```python
y_pred_unscaled = lr.predict(x_test)

y_pred_scaled = lr_scaled.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_unscaled = accuracy_score(y_pred_unscaled,y_test)
accuracy_scaled = accuracy_score(y_pred_scaled,y_test)
```

```python
print(f"Accuracy on unscaled dataframe = {accuracy_unscaled}.\nAccuracy on scaled dataframe ={accuracy_scaled}. ")
```
